# Notebook 01: Spatial Data Preprocessing & Coordinate Reference Systems (CRS)

**Project:** Optimal Electric Vehicle (EV) Charging Station Placement in Dhaka Metropolitan Area  
**Objective:** Ingest raw spatial datasets (OpenStreetMap road hierarchy, DPDC/DESCO 33/11kV substations, DTCA traffic counts, and RAJUK DAP zoning), transform Coordinate Reference Systems from WGS84 (`EPSG:4326`) to UTM Zone 46N (`EPSG:32646`), and construct spatial demand centroids.

```python
import os
import sys
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Ensure project root is in sys.path
module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)

from src.data_generator import generate_all_data

# Verify and load raw datasets
generate_all_data(module_path)
raw_dir = os.path.join(module_path, 'data', 'raw')
print('Datasets successfully verified in:', raw_dir)
```

## 1. Inspecting DPDC & DESCO Substation Inventory

```python
subs_df = pd.read_csv(os.path.join(raw_dir, 'dpdc_desco_substations.csv'))
print(f'Total Substations: {len(subs_df)}')
display(subs_df.head(10))
```

## 2. Inspecting OSM Road Network & DTCA Traffic Counts

```python
roads_gdf = gpd.read_file(os.path.join(raw_dir, 'osm_dhaka_roads.geojson'))
print(f'Road Network Segments: {len(roads_gdf)}')
print(f'Coordinate Reference System: {roads_gdf.crs}')
display(roads_gdf[['u', 'v', 'highway', 'maxspeed', 'lanes', 'length_m', 'pcu_per_hr', 'congestion_factor']].head())
```

## 3. Land Use Zoning & Flood Inundation Hazards (RAJUK DAP)

```python
landuse_gdf = gpd.read_file(os.path.join(raw_dir, 'rajuk_dap_landuse.geojson'))
fig, ax = plt.subplots(figsize=(9, 9))
landuse_gdf.plot(column='zone_type', legend=True, ax=ax, cmap='tab10', alpha=0.7, edgecolor='black')
plt.title('RAJUK Detailed Area Plan (DAP) Land Use & Exclusion Zones (Dhaka)', fontsize=13, fontweight='bold')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
