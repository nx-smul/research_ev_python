# Notebook 06: Comprehensive Result Visualization & Spatial Decision Maps

**Objective:** Render publication-grade static maps for the research outputs. Interactive map exploration is provided by the React/MapLibre web application and FastAPI service.

```python
import os
import sys
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)

from src.visualization.map_plots import plot_optimal_cs_locations
from src.visualization.pareto_front import find_knee_point_solution

cand_path = os.path.join(module_path, 'data', 'processed', 'candidate_sites_filtered.geojson')
cand_gdf = gpd.read_file(cand_path)

sub_path = os.path.join(module_path, 'data', 'raw', 'dpdc_desco_substations.csv')
subs_df = pd.read_csv(sub_path)

pareto_path = os.path.join(module_path, 'results', 'tables', 'optimal_solutions_pareto.csv')
pareto_df = pd.read_csv(pareto_path)
_, knee_sol = find_knee_point_solution(pareto_df)
selected_ids = knee_sol['selected_station_ids'].split(';')

# 1. Plot static optimal placement map
static_map_path = os.path.join(module_path, 'results', 'figures', 'optimal_cs_locations.png')
plot_optimal_cs_locations(cand_gdf, selected_ids, subs_df, static_map_path)

# Launch the interactive React/MapLibre dashboard from the repository root:
# Terminal 1: uvicorn api.app:app --reload --port 8000
# Terminal 2: cd web && npm run dev
print('Interactive map dashboard: start the FastAPI and web commands shown above.')
```
